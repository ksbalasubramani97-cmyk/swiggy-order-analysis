```python
# ============================================================
#              SWIGGY ORDER ANALYSIS PROJECT
#       Using NumPy, Pandas, Matplotlib & Seaborn
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("swiggy_orders.csv")

print("\n========== FIRST 5 ROWS ==========")
print(df.head())

print("\n========== DATASET INFORMATION ==========")
print(df.info())

print("\nRows and Columns:", df.shape)


# ============================================================
# 2. BASIC DATA ANALYSIS
# ============================================================

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

print("\n========== DUPLICATE ROWS ==========")
print(df.duplicated().sum())

print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())


# ============================================================
# 3. DATA CLEANING
# ============================================================

# Remove duplicates
df = df.drop_duplicates()

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Create Total Amount
df["Total_Amount"] = df["Quantity"] * df["Price"]

print("\n========== CLEANED DATA ==========")
print(df.head())


# ============================================================
# 4. TOTAL ORDERS
# ============================================================

total_orders = df["Order_ID"].nunique()

print("\n========== TOTAL ORDERS ==========")
print("Total Orders:", total_orders)


# ============================================================
# 5. TOTAL REVENUE
# ============================================================

total_revenue = df["Total_Amount"].sum()

print("\n========== TOTAL REVENUE ==========")
print("Total Revenue: ₹", round(total_revenue, 2))


# ============================================================
# 6. AVERAGE ORDER VALUE
# ============================================================

average_order_value = df["Total_Amount"].mean()

print("\n========== AVERAGE ORDER VALUE ==========")
print("Average Order Value: ₹",
      round(average_order_value, 2))


# ============================================================
# 7. TOP 10 RESTAURANTS
# ============================================================

top_restaurants = (
    df.groupby("Restaurant")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n========== TOP 10 RESTAURANTS ==========")
print(top_restaurants)


# ============================================================
# 8. TOP RESTAURANTS - BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_restaurants.values,
    y=top_restaurants.index
)

plt.title("Top 10 Restaurants by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Restaurant")

plt.tight_layout()
plt.show()


# ============================================================
# 9. ORDERS BY CITY
# ============================================================

city_orders = (
    df.groupby("City")["Order_ID"]
    .nunique()
    .sort_values(ascending=False)
)

print("\n========== ORDERS BY CITY ==========")
print(city_orders)


# ============================================================
# 10. ORDERS BY CITY - BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_orders.index,
    y=city_orders.values
)

plt.title("Orders by City")
plt.xlabel("City")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 11. REVENUE BY CITY
# ============================================================

city_revenue = (
    df.groupby("City")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== REVENUE BY CITY ==========")
print(city_revenue)


# ============================================================
# 12. REVENUE BY CITY - BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_revenue.index,
    y=city_revenue.values
)

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 13. MOST POPULAR FOOD ITEMS
# ============================================================

popular_food = (
    df.groupby("Food_Item")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n========== TOP 10 FOOD ITEMS ==========")
print(popular_food)


# ============================================================
# 14. POPULAR FOOD - BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=popular_food.values,
    y=popular_food.index
)

plt.title("Top 10 Most Ordered Food Items")
plt.xlabel("Quantity Ordered")
plt.ylabel("Food Item")

plt.tight_layout()
plt.show()


# ============================================================
# 15. FOOD CATEGORY ANALYSIS
# ============================================================

category_orders = (
    df.groupby("Category")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== FOOD CATEGORY ANALYSIS ==========")
print(category_orders)


# ============================================================
# 16. FOOD CATEGORY - PIE CHART
# ============================================================

plt.figure(figsize=(8, 6))

plt.pie(
    category_orders.values,
    labels=category_orders.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Orders by Food Category")

plt.show()


# ============================================================
# 17. MONTHLY REVENUE
# ============================================================

df["Month"] = df["Order_Date"].dt.month_name()

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

monthly_revenue = (
    df.groupby("Month")["Total_Amount"]
    .sum()
    .reindex(month_order)
)

print("\n========== MONTHLY REVENUE ==========")
print(monthly_revenue)


# ============================================================
# 18. MONTHLY REVENUE - LINE CHART
# ============================================================

plt.figure(figsize=(12, 6))

sns.lineplot(
    x=monthly_revenue.index,
    y=monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# 19. DELIVERY TIME ANALYSIS
# ============================================================

average_delivery = df["Delivery_Time"].mean()

print("\n========== DELIVERY TIME ==========")
print(
    "Average Delivery Time:",
    round(average_delivery, 2),
    "minutes"
)


# ============================================================
# 20. DELIVERY TIME DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Delivery_Time"],
    bins=20,
    kde=True
)

plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()


# ============================================================
# 21. RATING ANALYSIS
# ============================================================

average_rating = df["Rating"].mean()

print("\n========== CUSTOMER RATING ==========")
print("Average Rating:",
      round(average_rating, 2))


# ============================================================
# 22. RATING DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Rating"
)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Orders")

plt.show()


# ============================================================
# 23. PRICE VS RATING
# ============================================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Price",
    y="Rating"
)

plt.title("Price vs Customer Rating")
plt.xlabel("Food Price")
plt.ylabel("Rating")

plt.show()


# ============================================================
# 24. QUANTITY VS REVENUE
# ============================================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Total_Amount"
)

plt.title("Quantity vs Total Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.show()


# ============================================================
# 25. CORRELATION ANALYSIS
# ============================================================

numeric_columns = [
    "Quantity",
    "Price",
    "Total_Amount",
    "Delivery_Time",
    "Rating"
]

correlation = df[numeric_columns].corr()

print("\n========== CORRELATION ==========")
print(correlation)


# ============================================================
# 26. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()


# ============================================================
# 27. TOP 10 CUSTOMERS
# ============================================================

top_customers = (
    df.groupby("Customer_ID")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n========== TOP 10 CUSTOMERS ==========")
print(top_customers)


# ============================================================
# 28. PAYMENT METHOD ANALYSIS
# ============================================================

payment_analysis = (
    df.groupby("Payment_Method")["Order_ID"]
    .nunique()
    .sort_values(ascending=False)
)

print("\n========== PAYMENT METHOD ANALYSIS ==========")
print(payment_analysis)


# ============================================================
# 29. PAYMENT METHOD - BAR CHART
# ============================================================

plt.figure(figsize=(8, 5))

sns.barplot(
    x=payment_analysis.index,
    y=payment_analysis.values
)

plt.title("Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ============================================================
# 30. ORDER STATUS ANALYSIS
# ============================================================

order_status = (
    df["Order_Status"]
    .value_counts()
)

print("\n========== ORDER STATUS ==========")
print(order_status)


# ============================================================
# 31. ORDER STATUS - PIE CHART
# ============================================================

plt.figure(figsize=(7, 7))

plt.pie(
    order_status.values,
    labels=order_status.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Order Status Distribution")

plt.show()


# ============================================================
# 32. FINAL BUSINESS INSIGHTS
# ============================================================

top_city = (
    df.groupby("City")["Order_ID"]
    .nunique()
    .idxmax()
)

top_restaurant = (
    df.groupby("Restaurant")["Total_Amount"]
    .sum()
    .idxmax()
)

most_popular_food = (
    df.groupby("Food_Item")["Quantity"]
    .sum()
    .idxmax()
)

best_category = (
    df.groupby("Category")["Quantity"]
    .sum()
    .idxmax()
)

best_payment = (
    df.groupby("Payment_Method")["Order_ID"]
    .nunique()
    .idxmax()
)

print("\n")
print("=" * 60)
print("             SWIGGY ORDER ANALYSIS")
print("=" * 60)

print("Total Orders          :", total_orders)
print("Total Revenue         : ₹", round(total_revenue, 2))
print("Average Order Value   : ₹",
      round(average_order_value, 2))
print("Average Delivery Time :", round(average_delivery, 2),
      "minutes")
print("Average Rating        :", round(average_rating, 2))

print("-" * 60)

print("Top City              :", top_city)
print("Top Restaurant        :", top_restaurant)
print("Most Popular Food     :", most_popular_food)
print("Best Food Category    :", best_category)
print("Most Used Payment     :", best_payment)

print("=" * 60)
print("             ANALYSIS COMPLETED")
print("=" * 60)
```
